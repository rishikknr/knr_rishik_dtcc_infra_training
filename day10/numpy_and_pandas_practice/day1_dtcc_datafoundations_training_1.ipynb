```python
features = ['Slope', 'Rainfall', 'FA', 'Drainage']

for col in features:
    min_val = df[col].min()
    max_val = df[col].max()
    df[col + '_norm'] = (
        (df[col] - min_val) / (max_val - min_val)
        if max_val != min_val else 0
    )
```

```python
df['Risk_Score'] = (
    df['Slope_norm'] * 0.35 +
    df['Rainfall_norm'] * 0.35 +
    df['FA_norm'] * 0.20 +
    (1 - df['Drainage_norm']) * 0.10
) * 100
```

```python
df
```

```python
df['Risk_Level'] = pd.cut(
    df['Risk_Score'],
    bins=[-0.01, 25, 50, 75, 100],
    labels=['Low', 'Moderate', 'High', 'Very_High'],
    include_lowest=True
)

df[['Slope', 'Rainfall', 'Risk_Score', 'Risk_Level']].head()
```
